```python
import yfinance as yf
import json

try:
    # Lade Daten für Apple (AAPL)
    apple_data = yf.Ticker("AAPL")

    # Abrufe von Basisinformationen
    info = apple_data.info
    
    ticker = "AAPL"  # Da du den Ticker bereits kennst
    company_name = info.get('longName', 'Nicht verfügbar')
    market = info.get('market', 'Nicht verfügbar')
    exchange = info.get('exchange', 'Nicht verfügbar')
    category = info.get('quoteType', 'Nicht verfügbar')  # Oft als Kategorie verwendet
    sector = info.get('sector', 'Nicht verfügbar')

    print(f"Ticker: {ticker}")
    print(f"Name: {company_name}")
    print(f"Markt: {market}") ##oder category
    print(f"Börse: {exchange}")
    print(f"Kategorie: {category}")
    print(f"Sektor: {sector}")

except json.JSONDecodeError:
    print("Ein Decodierungsproblem ist aufgetreten. Bitte versuche es später erneut.")
except Exception as e:
    print(f"Ein Fehler ist aufgetreten: {e}")
```
